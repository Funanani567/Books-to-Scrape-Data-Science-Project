```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def visualize():

    df = pd.read_csv(
        "books_clean_dataset.csv"
    )

    sns.set_style("whitegrid")

    plt.figure(figsize=(8,5))
    sns.histplot(
        df["Price"],
        bins=20,
        kde=True
    )
    plt.title("Price Distribution")
    plt.show()

    plt.figure(figsize=(8,5))
    sns.countplot(
        x="Rating",
        data=df
    )
    plt.title("Rating Distribution")
    plt.show()

    plt.figure(figsize=(8,5))
    sns.boxplot(
        x="Rating",
        y="Price",
        data=df
    )
    plt.title("Price vs Rating")
    plt.show()

    plt.figure(figsize=(6,4))
    sns.heatmap(
        df[
            [
                "Price",
                "Rating",
                "Stock_Status",
                "Title_Length"
            ]
        ].corr(),
        annot=True,
        cmap="coolwarm"
    )

    plt.title("Correlation Matrix")
    plt.show()

if __name__ == "__main__":
    visualize()
```
