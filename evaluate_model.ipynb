```
import numpy as np
from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

from linear_regression_model import (
    train_model
)

def evaluate():

    model, X_test, y_test = train_model()

    predictions = model.predict(
        X_test
    )

    r2 = r2_score(
        y_test,
        predictions
    )

    mae = mean_absolute_error(
        y_test,
        predictions
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_test,
            predictions
        )
    )

    print("R² Score :", r2)
    print("MAE      :", mae)
    print("RMSE     :", rmse)

if __name__ == "__main__":
    evaluate()
```
