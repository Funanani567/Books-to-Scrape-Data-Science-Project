```
import pandas as pd
df = pd.read_csv("books_raw_dataset.csv")
def clean_data():

    df = pd.read_csv(
        "books_raw_dataset.csv"
    )

    rating_map = {
        "One": 1,
        "Two": 2,
        "Three": 3,
        "Four": 4,
        "Five": 5
    }

    df.drop_duplicates(
        inplace=True
    )

    df["Price"] = pd.to_numeric(
        df["Price"],
        errors="coerce"
    )

    df["Rating"] = df["Rating"].map(
        rating_map
    )

    df["Stock_Status"] = df["Stock"].apply(
        lambda x: 1 if "In stock" in x else 0
    )

    df["Title_Length"] = (
        df["Title"]
        .astype(str)
        .apply(len)
    )

    df.dropna(inplace=True)

    df.to_csv(
        "books_clean_dataset.csv",
        index=False
    )

    print("Dataset cleaned successfully.")

if __name__ == "__main__":
    clean_data()
```
